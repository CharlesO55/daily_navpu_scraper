```python
import pandas as pd
from pathlib import Path

def clean(input, save_file):
    df = pd.read_csv(input, 
                    header=None,
                    names=['date', 'value'])

    df = df.set_index('date').drop_duplicates()
    df['value'] = pd.to_numeric(df['value'].replace(',',''), errors='coerce')
    df = df.dropna()

    df.to_csv(save_file)
```

```python

BANK = "unionbank"
INPUT_FOLDER = Path(f"../../data/{BANK}")
OUTPUT_FOLDER = Path(f"../../data/navpu/{BANK}")

for file_path in INPUT_FOLDER.glob('*.csv'):
    save_file = OUTPUT_FOLDER / file_path.name
    clean(file_path, save_file)
    print(save_file)
```
